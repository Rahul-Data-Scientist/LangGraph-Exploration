# Some discussions on implementation of Long Term Memory in LangGraph

* In our AgenticAI notes, we have already discussed the concept of Long Term Memory (LTM) in detail. Here, we are going to discuss the implementation of LTM in LangGraph.
* In LangGraph, the concept of memory store is implemented by an abstract class called `BaseStore`.
* In the BaseStore abstract class, several methods are implemented which include get, search, put, etc. We will use these methods in this notebook.
* Several other classes inherit from this BaseStore. Some of these classes are:

    1. InMemoryStore → stores memory in RAM. Memory is destroyed after program's execution.

    2. PostgresStore → stores memory in Postgres database. Memory persists even after completion of program's execution. Used in production-grade applications.
    
    3. RedisStore → stores memory in Redis database.

```python
from langgraph.store.memory import InMemoryStore
```

```python
# create a store
store = InMemoryStore()
```

```python
# create a namespace
namespace = ("user", "u1")
```

# Explanation of Namespace

* If you think of a memory store as Google Drive, then a **namespace** is similar to a folder inside Google Drive.
* Just like folders organize files in a structured manner, namespaces organize memories in a structured manner.

### Example

* `namespace = ("user", "u1")`

This means:
- A **top-level folder** named `user` is created.
- Inside this folder, a **subfolder** named `u1` is created.
- This subfolder contains all memories of the user with `user_id = "u1"`.

### Other Possible Namespaces

1. `("user", "u1", "profile")`  
   - `user` → top-level folder  
   - `u1` → subfolder inside `user`  
   - `profile` → subfolder inside `u1`  
   - This folder stores **profile-related memories** of the user with `user_id = "u1"`, such as:
     - user's name
     - age
     - occupation

2. `("user", "u1", "preference")`  
   - `preference` is a subfolder under `("user", "u1")`.
   - This folder stores **user preferences** of the user with `user_id = "u1"`, such as:
     - user loves Python programming
     - user likes brief explanations of topics


# Creating Memories

```python
# Adding memories
store.put(namespace = namespace, key = "1", value = {"data": "User likes pizza"})
store.put(namespace = namespace, key = "2", value = {"data": "user prefers dark mode"})

# Har ek unique data item ko store karne ke liye ek unique key provide karna hota hai as you can see.
```

```python
# another namespace
namespace2 = ("user", "u2")
```

```python
# Adding memories
store.put(namespace = namespace2, key = "1", value = {"data": "User likes pasta"})
store.put(namespace = namespace2, key = "2", value = {"data": "User prefers grid style navigation"})
```

# Retrieving Memories

```python
# Retireving a specific memory
store.get(namespace = namespace, key = "1")
```

```python
store.get(namespace = namespace2, key = "1").value
```

```python
# Retireving all memories
items = store.search(namespace)
for item in items:
    print(item.value)
```

# Semantic search

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
embedding_model = OpenAIEmbeddings(model = "text-embedding-3-small")
```

```python
store = InMemoryStore(index = {"embed": embedding_model, "dims": 1536})  # By providing index, we are allowing semantic search
```

```python
namespace = ("users", "u1")
```

```python
store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User likes examples in Python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace, "8", {"data": "User is based in India"})
store.put(namespace, "9", {"data": "User likes real-world analogies"})
store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})
```

```python
items = store.search(namespace, query = "What is the user currently learning?", limit = 1)  # limit = n fetches top n items

for item in items:
    print(item.value)
```

```python
items = store.search(namespace, query = "What are user's preferences?", limit = 3)

for item in items:
    print(item.value)
```
